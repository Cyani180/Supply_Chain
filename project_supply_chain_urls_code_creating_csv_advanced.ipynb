```python
from bs4 import BeautifulSoup as bs
import requests
from lxml import etree





def get_reviews():
    #create empty lists:
    names, dates, stars, comments, headings, companies, invitations, rating_numbers_customer = [],[],[],[], [],[],[],[]
    #range_url=[]
    #for-loop to catch die reviews of pages 2 - 10, for example:

    for i in range (63,82):
        urls = [f'https://de.trustpilot.com/review/www.skatedeluxe.com?page={i}',
                f'https://de.trustpilot.com/review/www.kickz.com?page={i}',
                f'https://de.trustpilot.com/review/www.zalando.de?page={i}',
                f'https://de.trustpilot.com/review/www.deichmann.com?page={i}',
                f'https://de.trustpilot.com/review/fashionette.at?page={i}',
                f'https://de.trustpilot.com/review/freiluftkind.de?page={i}',
                f'https://de.trustpilot.com/review/www.adidas.com?page={i}',
                f'https://de.trustpilot.com/review/puma.com?page={i}',
                f'https://de.trustpilot.com/review/outfittery.de?page={i}',
                f'https://de.trustpilot.com/review/www.bonprix.de?page={i}']
         
        
        for url in urls:
            page=requests.get(url)
            soup=bs(page.content, 'lxml')
            
            reviews = soup.find_all('div', class_='styles_cardWrapper__kOLEb styles_show__qAseP')
        
        
        #fill lists wih content:
            for content in reviews:
                name = content.find('span', attrs={'class':'typography_heading-xs__osRhC typography_appearance-default__t8iAq' ,'data-consumer-name-typography':'true'}).text.strip()
                names.append(name)

                date = content.find('div', class_= 'typography_body-m__k2UI7 typography_appearance-subtle__PYOVM').text.strip()
                dates.append(date)

            #some reviews are without comments, so there are NaN, and we Python get an error!!!
            #We need a loop, to fill empty comments
            
                comment =content.find('p', attrs={'class':'typography_body-l__v5JLj typography_appearance-default__t8iAq','data-service-review-text-typography':'true'})
            

                if comment is None:
                    comments.append("No comment")
                
                else:
                    comments.append(comment)
                
            
                star=content.find('div', class_='styles_reviewHeader__PuHBd')['data-service-review-rating']
                stars.append(star)

                heading =content.find('h2', attrs = {'class': 'typography_heading-xs__osRhC typography_appearance-default__t8iAq','data-service-review-title-typography':'true'}).text.strip()
                headings.append(heading)

                rating_number_customer= content.find('div', class_='styles_consumerExtraDetails__TylYM')['data-consumer-reviews-count']
                rating_numbers_customer.append(rating_number_customer)

                invitation = content.find('svg', class_='icon_icon__RdICC icon_appearance-inherit__MeZek')
                if invitation is None:
                    invitations.append("No Invitation")
                else:
                    invitations.append("Auf Einladung")
            

                company = soup.find('span', class_='typography_display-s__pKPhT typography_appearance-default__t8iAq title_displayName__OBSVU').text.strip()
                companies.append(company)

                

                
                
                
    
    import pandas as pd
    
    df_trustpilot = pd.DataFrame(list(zip(companies, names,rating_numbers_customer, headings, comments, stars,invitations, dates)),
               columns =['Company', 'Name','Rating_number_customer', 'Heading','Comment','Stars','Invitation', 'Dates'])

            
    print(df_trustpilot.head())
    df_trustpilot.to_csv('supply_chain_project_trustpilot_advanced_7.csv', index=False)
    df_trustpilot.info()
        
get_reviews()

```
